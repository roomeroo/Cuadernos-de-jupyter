# Reglas lógicas que debe cumplir el ordenador
Para que el programa sea válido (ya sea que automatices la solución o crees un juego interactivo), el algoritmo o las funciones de control deben verificar estrictamente estas tres condiciones básicas por cada movimiento:

- Solo puedes mover el disco superior de una varilla.
- Solo puedes mover un disco a la vez.
- Nunca puedes poner un disco mayor sobre uno menor.


El número de varillas **siempre** es 3

```python
tiempo_consola = 0.5


numero_discos = 5 #int(input("Numero de discos: ")) # Reto incial 3

if(numero_discos <= 0): numero_discos = 1
```

## Creo la clase Varilla
Como no existe el tipo de dato pila, finjo que existe usando por detras un array.

```python
class Varilla:
  def __init__(self, nombre, numero_discos = 1, es_inicial = False):
    self.nombre = nombre
    self._discos = []
    if(es_inicial): self.iniciar_varilla(numero_discos)

  def peek(self):
    if(not self.esta_vacia()):
      return self._discos[-1] # Devuelvo el último sin tocar nada
    else:
      print("Está vacía")
      return 0 # Siempre va a ser mas pequeño que caulquier disco

  def esta_vacia(self):
    if(len(self._discos) == 0):
      return True
    else: return False

  def desapilar(self):
    if(len(self._discos) > 0):
      self._discos.pop()

  def apilar(self, varilla_destino):
    if(self.esta_vacia()): return -1 # Está vacía no se produce ningun cambio
    if(self.peek() > varilla_destino.peek() and not varilla_destino.esta_vacia()): # Compruebo por si acaso
      return -1 # No se puede apilar un disco mas grande sobre uno mas pequeño
    else:
      varilla_destino._discos.append(self.peek()) # Copio el ultimo de la varilla origen en la varilla objetivo
      self.desapilar() # Borro el último

  def iniciar_varilla(self, numero_discos):
    for i in range(numero_discos, 0, -1):
      self._discos.append(i)

  def pintar_varilla(self):
    ancho = 2 * numero_discos + 1                              # ancho del disco más grande
    for nivel in range(numero_discos, -1, -1):                 # de arriba abajo
      if nivel < len(self._discos):
        pieza = "█" * (2 * self._discos[nivel] + 1)
      else:
        pieza = "│"                                # hueco: solo el palo
      print(pieza.center(ancho))
    print("▄" * ancho)                             # la base
    print(self.nombre.center(ancho))
    print()

```

## Creo las varillas con toda la calma del mundo

```python
varillas = [
  Varilla(nombre="A", numero_discos=numero_discos, es_inicial=True),
  Varilla("B", numero_discos=numero_discos),
  Varilla("C", numero_discos=numero_discos),
]
```

## Método para saber si he ganado ya.

```python
def esta_resuelto(*varillas):
  for varilla in varillas[0:-1]:
    if(not varilla.esta_vacia()): return False
    return True
```

## Método para pintar las varillas

```python
def pintar_varillas(varillas):
  for varilla in varillas:
    varilla.pintar_varilla()
```

## Método para esperar y limpiar consola

```python
import time
from IPython.display import clear_output
def espera_limpia():
  time.sleep(tiempo_consola)
  clear_output(wait=True)
```

Pero luego cuando queremos quitarnos los discos de encima, el origen sigue siendo el mismo, pero el destino es B y el auxiliar es C.
Luego cuando queremos seguir completanddo, C
